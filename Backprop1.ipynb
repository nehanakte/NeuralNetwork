```python
import math

# Sigmoid and derivative
def sigmoid(x):
    return 1 / (1 + math.exp(-x))

def sigmoid_derivative(y):
    return y * (1 - y)

# -------- GIVEN VALUES (from diagram) --------
x1 = 0.35
x2 = 0.9
target = 0.5
lr = 1

# Weights
w13 = 0.1
w14 = 0.4
w23 = 0.8
w24 = 0.6
w35 = 0.3
w45 = 0.9

# -------- FORWARD PASS (Before Training) --------
# Hidden layer
h3_in = x1*w13 + x2*w23
h3 = sigmoid(h3_in)

h4_in = x1*w14 + x2*w24
h4 = sigmoid(h4_in)

# Output layer
o5_in = h3*w35 + h4*w45
o5 = sigmoid(o5_in)

loss_before = 0.5 * (target - o5) ** 2

print("Before Training")
print("Output:", round(o5, 4))
print("Loss:", round(loss_before, 4))
print("-" * 40)

# -------- BACKPROPAGATION --------
# Output neuron delta
delta_o5 = (target - o5) * sigmoid_derivative(o5)

# Hidden layer deltas
delta_h3 = sigmoid_derivative(h3) * delta_o5 * w35
delta_h4 = sigmoid_derivative(h4) * delta_o5 * w45

# -------- UPDATE WEIGHTS --------
w35 += lr * delta_o5 * h3
w45 += lr * delta_o5 * h4

w13 += lr * delta_h3 * x1
w23 += lr * delta_h3 * x2
w14 += lr * delta_h4 * x1
w24 += lr * delta_h4 * x2

# -------- FORWARD PASS (After Training) --------
h3 = sigmoid(x1*w13 + x2*w23)
h4 = sigmoid(x1*w14 + x2*w24)
o5 = sigmoid(h3*w35 + h4*w45)

loss_after = 0.5 * (target - o5) ** 2

print("After Training")
print("Output:", round(o5, 4))
print("Loss:", round(loss_after, 4))
```
