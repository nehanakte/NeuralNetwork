```python
import math

def sigmoid(x):
    return 1 / (1 + math.exp(-x))

def sigmoid_derivative(y):
    return y * (1 - y)

# Inputs
x1, x2, x3 = 1, 0, 1
target = 1
lr = 1

# Weights
w14, w15 = 0.2, -0.3
w24, w25 = 0.4, 0.1
w34, w35 = -0.5, 0.2
w46, w56 = -0.3, -0.2

# Biases
b4, b5, b6 = 0.4, 0.2, 0.1

# -------- Forward Pass (Before Training) --------
h4_in = x1*w14 + x2*w24 + x3*w34 + b4
h4 = sigmoid(h4_in)

h5_in = x1*w15 + x2*w25 + x3*w35 + b5
h5 = sigmoid(h5_in)

o6_in = h4*w46 + h5*w56 + b6
o6 = sigmoid(o6_in)

loss_before = 0.5 * (target - o6) ** 2

print("Before Training")
print("Output:", o6)
print("Loss:", loss_before)

# -------- Backpropagation --------
delta_o6 = (target - o6) * sigmoid_derivative(o6)

delta_h4 = sigmoid_derivative(h4) * delta_o6 * w46
delta_h5 = sigmoid_derivative(h5) * delta_o6 * w56

# -------- Update Weights --------
w46 = w46 + lr * delta_o6 * h4
w56 = w56 + lr * delta_o6 * h5
b6  = b6  + lr * delta_o6

w14 = w14 + lr * delta_h4 * x1
w24 = w24 + lr * delta_h4 * x2
w34 = w34 + lr * delta_h4 * x3
b4  = b4  + lr * delta_h4

w15 = w15 + lr * delta_h5 * x1
w25 = w25 + lr * delta_h5 * x2
w35 = w35 + lr * delta_h5 * x3
b5  = b5  + lr * delta_h5

# -------- Print Updated Weights --------
print("\nUpdated Weights")
print("w14:", w14)
print("w24:", w24)
print("w34:", w34)
print("w15:", w15)
print("w25:", w25)
print("w35:", w35)
print("w46:", w46)
print("w56:", w56)

# -------- Forward Pass (After Training) --------
h4 = sigmoid(x1*w14 + x2*w24 + x3*w34 + b4)
h5 = sigmoid(x1*w15 + x2*w25 + x3*w35 + b5)
o6 = sigmoid(h4*w46 + h5*w56 + b6)

loss_after = 0.5 * (target - o6) ** 2

print("\nAfter Training")
print("Output:", o6)
print("Loss:", loss_after)
```
